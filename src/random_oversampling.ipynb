### Models with Random Oversampling and Features from Baselines 

#### Random Oversampling 

https://imbalanced-learn.org/stable/references/generated/imblearn.over_sampling.RandomOverSampler.html#imblearn.over_sampling.RandomOverSampler.fit_resample

```python
# do random oversampling 
ros = RandomOverSampler(random_state=42)
X_train_res, y_train_res = ros.fit_resample(X_train, y_train)
```

```python
# check that the data is balanced now 
c = Counter(y_train_res)
c
```

```python
# get the result from running all baselines with the balanced dataset
result = run_all_baselines(X_train_res, X_test, y_train_res, y_test)
```
